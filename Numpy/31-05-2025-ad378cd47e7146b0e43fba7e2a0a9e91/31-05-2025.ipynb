# Questions

- A 4x4 identity matrix
- 10 evenly spaced numbers between 1 and 100
- A 2x2 array with random values between 0 and 1


Analogy: Boxes in boxes
Imagine you have a gift box (array) with shape:

Case 1: [1, 2, 3]
➡️ 1 box with 2 smaller boxes with 3 items
You can remove the outer box (because there's only 1), and still have the same contents. ✅

So,
