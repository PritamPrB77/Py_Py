```python
print(a-b)
print(a*b)
print(a/b)
print(a>b)
ans=a>b
print(f"{type (ans[0])}-")
print(a<b)
print(a==b)
print(a|b)
print(a&b)
print(a!=b)
```

```python
x=np.array([False, False, True]);
y=np.array([True, False, True]);
print(np.logical_and(x,y))
print(np.logical_or(x,y))
print(np.logical_not(x,y))
```

# bitwise Operator

```python
a=np.array([1,2,3])
b=np.array([4,5,6])

print(np.bitwise_xor(a,b))
print(np.bitwise_and(a,b))

```
