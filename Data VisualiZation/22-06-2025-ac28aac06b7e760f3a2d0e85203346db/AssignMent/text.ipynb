```python
# 📦 Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 📥 Load the dataset
df = pd.read_excel("data/Online Retail.xlsx")

# 🧼 Clean the data (drop nulls in InvoiceNo or Country)
df = df.dropna(subset=["InvoiceNo", "Country"])

# 🔢 Count unique InvoiceNo (transactions) per country
country_transactions = df.groupby("Country")["InvoiceNo"].nunique()

# 📈 Sort and get Top 10 countries
top_10 = country_transactions.sort_values(ascending=False).head(10)

# 📊 Plot
plt.figure(figsize=(10, 5))
sns.barplot(y=top_10.values, x=top_10.index, palette="Blues_d")
plt.title("Top 10 Countries by Number of Transactions")
plt.xlabel("Number of Unique Transactions")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```
