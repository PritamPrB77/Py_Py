```python
num = int(input("Enter a 4-digit number: "))
sum_digits = sum([int(digit) for digit in str(num)])
print("Sum of digits:", sum_digits)
```

```python
num = int(input("Enter an integer: "))

if num & 1 == 0: print("Even number")
else:
    print("Odd number")
```

```python
while True:
    global num 
    num=int(input("Enter a positive number: "))
    if num > 0:
        break
    print("Please enter a positive number!")

if num % 2 == 0:
    print("Even → Result:", num * 2)
else:
    print("Odd → Result:", num ** 2)
```

```python
whiteRow=[2,4,6,8]
whiteCol=['b','d','f','h']
pos=input("Enter position: ")
row=int(pos.split(' ')[0])
col=pos.split(' ')[1]
if (row in whiteRow and col in whiteCol) :
    print("White Square")
```

```python
count = 0
for i in range(100, 1001):
    if i % 5 == 0 or i % 6 == 0:
        print(i, end=" ;;")
        count += 1
        if count % 10 == 0:
            print()
```
